# Lesson 05: Modular Code

## 1. What Is Modular Code?

**Modular code** means dividing a large program into smaller parts.

Each part performs a specific task, and all parts work together to complete the program.

### Example: School Result System

- One part calculates marks.
- One part calculates percentage.
- One part displays the result.
- One part prints the report card.

**Simple meaning:**  
Modular code = one large program divided into smaller, manageable parts.

---

## 2. What Is a Function?

A **function** is a block of code designed to perform a particular task.

A function can be used again whenever needed. This helps us reuse code and avoid repetition.

### Examples of Function Tasks

- Calculate marks
- Display a message
- Add two numbers
- Print a report

---

## 3. Function Definition

A **function definition** means creating a function.

```python
def welcome():
    print("Welcome to school")
```

### Parts of the Function

- `def` — tells Python that a function is being created.
- `welcome` — name of the function.
- `()` — contains parameters, if any.
- `print(...)` — task performed by the function.

> Creating a function does not run it automatically.

---

## 4. Function Call

A **function call** activates or executes a function.

```python
def welcome():
    print("Welcome to school")

welcome()
```

### Output

```text
Welcome to school
```

The function runs only after we call it using:

```python
welcome()
```

---

## 5. Parameter

A **parameter** is a variable that receives information passed to a function.

```python
def welcome(name):
    print("Welcome", name)

welcome("Ram")
```

### Output

```text
Welcome Ram
```

### Understanding the Example

- `name` is the parameter.
- `"Ram"` is the value passed to the function.
- The function uses the value stored in `name`.

---

## 6. Understanding the Textbook Activity

The textbook activity displays the stock levels of products **A, B, C, D and E** using bars.

The program uses a function named `DisplayBar`.

### Parameters of `DisplayBar`

The function has three parameters:

| Parameter | Purpose |
|---|---|
| Product Letter | Identifies the product: A, B, C, D or E |
| Symbol | Represents stock using a symbol |
| Quantity | Represents the available stock |

### Example

```python
def DisplayBar(product, symbol, quantity):
    print(product, symbol * quantity)

DisplayBar("A", "*", 5)
```

### Important Rule

Parameters must be passed in the **same order** as shown in the function definition:

```python
def DisplayBar(product, symbol, quantity):
```

The call:

```python
DisplayBar("A", "*", 5)
```

means:

- `"A"` is received by `product`.
- `"*"` is received by `symbol`.
- `5` is received by `quantity`.

## 7. Same Addition Program: Without and With a Function

### A. Addition Without a Function

```python
number1 = 10
number2 = 5

sum = number1 + number2
print("Sum:", sum)
```

### Output

```text
Sum: 15
```

### How It Works

1. `number1` stores `10`.
2. `number2` stores `5`.
3. Both numbers are added using `number1 + number2`.
4. The answer is stored in `sum`.
5. `print()` displays the answer.

> This program performs the sum directly. It does not use a function.

---

### B. Addition With a Function

```python
def add_numbers(number1, number2):
    sum = number1 + number2
    print("Sum:", sum)

add_numbers(10, 5)
```

### Output

```text
Sum: 15
```

### How It Works

1. `def add_numbers(number1, number2):` creates the function.
2. `number1` and `number2` are parameters.
3. The function adds the two numbers.
4. `add_numbers(10, 5)` calls the function.
5. `10` is passed to `number1`.
6. `5` is passed to `number2`.

### Difference Between the Two Programs

| Without Function | With Function |
|---|---|
| Code is written and used once | Code is kept inside a reusable function |
| Values are used directly | Values are passed through parameters |
| Harder to reuse | Easy to reuse with different numbers |
| Suitable for a small one-time task | Suitable for repeated tasks |

### Reusing the Function

The function can be called again with different values:

```python
add_numbers(20, 30)
add_numbers(7, 8)
add_numbers(100, 250)
```

### Output

```text
Sum: 50
Sum: 15
Sum: 350
```

> The function performs the same task but can be reused several times.

---

## 8. Why Use Functions?

Functions help us:

1. Divide a large program into smaller parts.
2. Reuse code.
3. Make programs easier to understand.
4. Find and correct mistakes easily.
5. Give each task a clear name and structure.

# How to Test a Program Using Break, Step and Watch

## 1. What Is Debugging?

**Debugging** means finding and correcting mistakes in a program.

Special tools called **debugging tools** help us examine the program while it is running.

The most important debugging tools are:

1. Breakpoint
2. Step
3. Watch

In many programs, the name **Break** means **Breakpoint** or **Pause**.

---

## 2. What Is a Breakpoint?

A **breakpoint** is a stopping point placed in the program.

When the program reaches that line, it pauses. This allows us to check the values and find mistakes.

### Example Program

```python
def add_numbers(number1, number2):
    sum = number1 + number2
    print("Sum:", sum)

add_numbers(10, 5)
```

### How to Use a Breakpoint

1. Open the Python program in VS Code.
2. Move the mouse to the left side of a line number.
3. Click beside the line number.
4. A red dot appears. This red dot is the breakpoint.
5. Click **Run and Debug** or press `F5`.
6. The program will stop at the breakpoint.

### Suggested Breakpoint

Place a breakpoint on:

```python
sum = number1 + number2
```

The program pauses before or after this line is executed, depending on VS Code's execution position.

### Why Use a Breakpoint?

A breakpoint helps us:

- Stop the program at an important line.
- Check the values of variables.
- Find where a wrong value is created.
- Avoid running the whole program repeatedly.

### Example: Finding a Logic Error

```python
def add_numbers(number1, number2):
    sum = number1 * number2
    print("Sum:", sum)

add_numbers(10, 5)
```

Place a breakpoint at:

```python
print("Sum:", sum)
```

The program pauses before the answer is printed. We can then notice that `sum` contains `50` instead of `15`.

---

## 3. What Is Step?

**Step** means running the program line by line.

It helps us see exactly what happens in each line.

### Different Step Commands

| Button or Key | Name | Meaning |
|---|---|---|
| `F10` | Step Over | Run the current line and move to the next line |
| `F11` | Step Into | Enter inside a function to examine it line by line |
| `Shift + F11` | Step Out | Leave the current function and return to the calling line |
| `F5` | Continue | Run until the next breakpoint |

> In some VS Code displays, these buttons are shown as Continue, Step Over, Step Into and Step Out.

---

### Step Over

**Step Over** runs the next line but does not enter inside a function.

```python
number1 = 10
number2 = 5

sum = number1 + number2
print("Sum:", sum)
```

Press `F10` four times:

1. Store `10` in `number1`.
2. Store `5` in `number2`.
3. Calculate `sum = 15`.
4. Print `Sum: 15`.

Use **Step Over** when:

- You do not need to enter a function.
- You only want to observe the current line.

---

### Step Into

**Step Into** enters a function and runs it line by line.

```python
def add_numbers(number1, number2):
    sum = number1 + number2
    print("Sum:", sum)

add_numbers(10, 5)
```

Place a breakpoint on:

```python
add_numbers(10, 5)
```

Press `F11` or **Step Into**.

The debugger moves inside the function:

```python
def add_numbers(number1, number2):
```

Now we can step through:

```python
sum = number1 + number2
print("Sum:", sum)
```

Use **Step Into** when:

- You want to examine the statements inside a function.
- You need to find a mistake in the function.

---

### Step Out

**Step Out** leaves the current function and returns to the line that called it.

Use **Step Out** when:

- You no longer need to check the function line by line.
- You want to continue testing the rest of the program.

---

## 4. What Is Watch?

A **watch** allows us to watch a variable or expression while the program runs.

It shows how values change at each step.

### How to Add a Watch

1. Start the program in Debug mode.
2. Pause it using a breakpoint.
3. Open the **Run and Debug** panel.
4. Find the **Watch** section.
5. Click the `+` button.
6. Enter a variable name or expression, such as:
   - `number1`
   - `number2`
   - `sum`
   - `number1 + number2`

The Watch panel will display the current value.

---

### Watch Example

```python
number1 = 10
number2 = 5

sum = number1 + number2
print("Sum:", sum)
```

Add these watches:

- `number1`
- `number2`
- `sum`
- `number1 + number2`

| Variable or Expression | Value |
|---|---:|
| `number1` | 10 |
| `number2` | 5 |
| `sum` | 15 |
| `number1 + number2` | 15 |

A watch makes it easy to see when a variable contains a wrong value.

---

## 5. Complete Testing Example

```python
def calculate_average(mark1, mark2, mark3):
    total = mark1 + mark2 + mark3
    average = total
    return average

answer = calculate_average(80, 90, 70)
print("Average:", answer)
```

### Steps to Test This Program

1. Place a breakpoint on:

   ```python
   total = mark1 + mark2 + mark3
   ```

2. Run the program using `F5`.
3. Add watches for:
   - `mark1`
   - `mark2`
   - `mark3`
   - `total`
   - `average`

4. Step through the program using `F10`.
5. Check the watch values.

### Expected Values

| Variable | Expected Value |
|---|---:|
| `mark1` | 80 |
| `mark2` | 90 |
| `mark3` | 70 |
| `total` | 240 |
| `average` | 80 |

The incorrect code has:

```python
average = total
```

It should be:

```python
average = total / 3
```

This is a **logic error** because the program adds the marks but does not divide the total by 3.

---

## 6. How the Three Tools Work Together

| Tool | Purpose | Main Question Answered |
|---|---|---|
| **Break/Breakpoint** | Pause the program at a selected line | Where should the program stop? |
| **Step** | Run the program one line or function at a time | What happens next? |
| **Watch** | Display and monitor variable values | What value is stored now? |

### Best Testing Method

1. Place a breakpoint near the suspicious line.
2. Run the program until the breakpoint.
3. Add watches for important variables.
4. Use **Step Over** to observe each line.
5. Use **Step Into** to enter a function.
6. Correct the line if the watch value is wrong.
7. Run the program again to test the result.

---

## 7. Short Questions and Answers

### Q1. What is a breakpoint?

**Answer:** A breakpoint pauses a program at a selected line so that we can examine it.

### Q2. What is Step Over?

**Answer:** Step Over runs one line at a time without entering a function.

### Q3. What is Step Into?

**Answer:** Step Into enters a function so we can examine its statements one line at a time.

### Q4. What is Step Out?

**Answer:** Step Out leaves the current function and returns to the calling line.

### Q5. What is a watch?

**Answer:** A watch shows the current value of a variable or expression while the program is running.

### Q6. Which key starts a program in Debug mode in VS Code?

**Answer:** Press `F5`.

### Q7. Which key is commonly used for Step Over?

**Answer:** Press `F10`.

### Q8. Which key is commonly used for Step Into?

**Answer:** Press `F11`.

---

## 8. Quick Revision

- **Breakpoint:** Stops the program at a selected line.
- **Continue:** Runs the program until the next breakpoint.
- **Step Over:** Runs the next line.
- **Step Into:** Goes inside a function.
- **Step Out:** Leaves the current function.
- **Watch:** Displays current variable values.

**Remember:** A breakpoint stops the program, Step controls line-by-line movement, and Watch displays values during testing.

# Testing for Functional, Operational and Logic Errors

## 1. What Is Testing?

**Testing** means running a program and checking whether it gives the correct and expected result.

### Purpose of Testing

Testing helps us:

- Find mistakes in a program
- Check whether the program performs its task correctly
- Check whether the program works in different conditions
- Improve the quality and reliability of a program

---

## 2. Functional Error

A **functional error** occurs when the program does not perform the function or task that it was designed to perform.

The program may run, but it gives an incorrect or missing result.

### Simple Meaning

**Expected work is not performed correctly.**

### Example: Addition Program

The program should calculate $10 + 5$.

```python
number1 = 10
number2 = 5

result = number1 - number2
print("Sum:", result)
```

### Output

```text
Sum: 5
```

### What Is Wrong?

The program should add the numbers, but it subtracts them.

- **Expected result:** $10 + 5 = 15$
- **Actual result:** $10 - 5 = 5$

This is a **functional error** because the addition function does not perform its required task.

### How to Test for a Functional Error

1. Read the requirement or expected function.
2. Give valid input to the program.
3. Calculate the correct answer manually.
4. Compare the expected result with the actual result.

| Test Input | Required Result | Actual Result | Status |
|---|---:|---:|---|
| 10 and 5 | 15 | 5 | Functional error found |
| 20 and 30 | 50 | 50 | Correct |

---

### Functional Test for a Function

```python
def add_numbers(number1, number2):
    return number1 + number2

print(add_numbers(10, 5))
print(add_numbers(20, 30))
```

### Expected Results

```text
15
50
```

The function is tested using different inputs to check whether it always performs the required task.

---

## 3. Operational Error

An **operational error** occurs when the program does not work properly in its real operating environment or when it receives unexpected input or conditions.

### Simple Meaning

**The program has difficulty running or performing its task under real conditions.**

Operational errors may include:

- Program crashing
- Wrong response to invalid input
- File cannot be opened
- Incorrect data type used
- Device or internet connection unavailable
- Not enough memory

### Example 1: Invalid Data Type

```python
number1 = input("Enter first number: ")
number2 = input("Enter second number: ")

result = number1 + number2
print("Sum:", result)
```

If the user enters `10` and `5`, the output may be:

```text
105
```

This happens because `input()` receives text values. Two text values are joined instead of being added.

### Operationally Correct Version

```python
number1 = int(input("Enter first number: "))
number2 = int(input("Enter second number: "))

result = number1 + number2
print("Sum:", result)
```

Now the input is converted into integer values before addition.

### Example 2: Unexpected Text Input

```python
number1 = int(input("Enter first number: "))
```

If the user enters `hello`, the program may stop with an error because `hello` is text, not an integer.

A safer program can display a message:

```python
try:
    number1 = int(input("Enter first number: "))
    number2 = int(input("Enter second number: "))

    print("Sum:", number1 + number2)
except ValueError:
    print("Please enter whole numbers only.")
```

### How to Test for Operational Errors

Test the program using:

- Valid inputs
- Invalid inputs
- Empty inputs
- Very large inputs
- Decimal or text values
- Missing files
- Slow or disconnected systems
- Different devices or operating conditions

### Operational Test Table

| Test Condition | Input | Expected Result |
|---|---|---|
| Valid input | 10 and 5 | 15 |
| Invalid text input | Hello | Please enter whole numbers only |
| Empty input | Press Enter | A clear error message |
| Large numbers | 1000 and 2000 | 3000 |

> Operational testing checks whether the program works safely and correctly under different real-world conditions.

---

## 4. Logic Error

A **logic error** occurs when the program has no syntax or running error, but it follows incorrect reasoning.

The program runs successfully, but its result is wrong.

### Simple Meaning

**The program runs, but it calculates or works in the wrong way.**

### Main Features of a Logic Error

- The program usually starts and runs.
- It does not normally stop with an error message.
- The answer or process is incorrect.
- It is often found by comparing output with the correct answer.

### Example 1: Addition Logic Error

```python
number1 = 10
number2 = 5

sum = number1 * number2
print("Sum:", sum)
```

### Output

```text
Sum: 50
```

### Mistake

The program uses multiplication (`*`) instead of addition (`+`).

- **Expected result:** $10 + 5 = 15$
- **Actual result:** $10 × 5 = 50$

The error is in the **program logic**.

### Correct Logic

```python
sum = number1 + number2
print("Sum:", sum)
```

### Output

```text
Sum: 15
```

---

### Example 2: Average Logic Error

A student marks are 80, 90 and 70. The average should be:

$$\frac{80+90+70}{3}=80$$

```python
marks1 = 80
marks2 = 90
marks3 = 70

average = marks1 + marks2 + marks3
print("Average:", average)
```

### Output

```text
Average: 240
```

### Mistake

The program adds the marks but does not divide by 3.

### Correct Logic

```python
average = (marks1 + marks2 + marks3) / 3
print("Average:", average)
```

### Output

```text
Average: 80.0
```

---

## 5. How Functional, Operational and Logic Errors Are Different

| Type of Error | Meaning | Example | Does the Program Run? |
|---|---|---|---|
| **Functional error** | The program does not perform the required function correctly | Addition program subtracts instead of adding | Often yes |
| **Operational error** | The program fails under real or unexpected operating conditions | Invalid text or missing file causes a crash | May stop or fail |
| **Logic error** | The program runs but follows incorrect reasoning | Average is not divided by the number of subjects | Yes, but output is wrong |

### Step 5: Compare Results

Compare:

- Check messages and conditions.
